```python
import moku_util as mu
from moku.instruments import Datalogger

# The moku_util module loads these libraries for internal methods,
# but you may want to run alternative code that uses them directly.
import matplotlib.pyplot as plt
import numpy as np
```

#### Connect to the Moku:Go via command-line, specifically activating the Datalogger functionality

```python
## MAKE SURE TO GET THE CORRECT IP ADDRESS FOR YOUR OWN
## MOKU:GO DEVICE. THIS IS JUST AN EXAMPLE. YOU MAY CONSIDER
## USING THE DEVICE SERIAL NUMBER INSTEAD, WHICH CAN BE 
## FOUND BY RUNNING "mokucli list" IN THE TERMINAL.
ip_address = "mokugo-000000.local"
ip_address = "[fe80::7269:79ff:feb9:668e%17]"
logger = Datalogger(ip_address, force_connect=True)
```

#### Setup the datalogger with default settings for input range and impedance etc.

You can setup single-channel acquisition which allows full use of the 1~MHz sampling rate, or you can setup two-channel acquisition, which cuts the maximum sample rate in half, i.e. 500~kHz. The example notebook here uses a two channel logger.

You shouldn't need to change any of the default arguments, but if you're interested, you could open the `moku_util` module code and investigate other options. 

```python
mu.setup_two_channel_logger(logger)
```

#### Acquire some data

Make sure to change the `savedir` to specify where you want the data stored locally, as well as a marginally descriptive filename so you know what data you took. Record operating conditions and corresponding filename in your lab notebook.

```python
sampling_frequency = 500000.0    # samples per second
duration = 10.0                  # duration of acquisition in seconds
delay = 0                        # optional acquisition delay in seconds

# Change this to your desired save directory. The .li file will 
# be saved in this directory with a name that includes your specified
# filename_prefix and a timestamp set by the DataLogger.
savedir = '../example_data'

# Log data from the Moku:Go and download it locally to your computer.
filename = mu.log_data(
    logger,
    sampling_frequency,
    duration,
    delay=delay,
    filename_prefix="example_file",
    save_directory=savedir,
    # use_trigger=True,
    # trigger_source="Input1",
    # trigger_level=0.25,
)
```

The full path of the data you just saved should be printed as output to the above cell. You can copy this filename into subsequent notebooks for analysis.

#### Make sure to ``disconnect'' from the Moku:Go once you're done acquiring data

This is left as a separate cell that you can optionally reacquire data to get multiple files for later analysis before disconnecting. You can also re-initialize the Moku:Go and collect more data if you disconnect prematurely.

```python
# This function disconnects the device so it can be used by the desktop app
logger.relinquish_ownership()
```

#### Let's quickly check our data to make sure it looks as expected

```python
# Load the data using the filename returned by the log_data function.
df = mu.DataFile(filename)
```

```python
%matplotlib widget
fig, ax = df.plot_data(input=[0,1], microseconds=True, xlim=(0,10000))
```
